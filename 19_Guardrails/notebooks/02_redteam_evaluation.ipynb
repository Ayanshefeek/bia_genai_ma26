```python
import pandas as pd; df=pd.read_csv('../redteam_results.csv'); df
```

```python
import matplotlib.pyplot as plt
unsafe=df[~df['overall_safe']]
plt.barh(unsafe['prompt'], unsafe['semantic_score'])
plt.xlabel('Semantic Risk Score')
plt.title('Unsafe Prompts in Red-Team Drill')
plt.show()
```
