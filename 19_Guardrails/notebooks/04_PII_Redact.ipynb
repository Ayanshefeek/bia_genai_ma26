```python
import re
from typing import Dict, Tuple

PII_PATTERNS = {
    "EMAIL": re.compile(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"
    ),

    "PHONE": re.compile(
        r"\b(?:\+?\d{1,3}[-.\s]?)?(?:\d{10}|\d{3}[-.\s]?\d{3}[-.\s]?\d{4})\b"
    ),

    "AADHAAR": re.compile(
        r"\b\d{4}\s?\d{4}\s?\d{4}\b"
    ),

    "PAN": re.compile(
        r"\b[A-Z]{5}[0-9]{4}[A-Z]\b"
    ),

    "CREDIT_CARD": re.compile(
        r"\b(?:\d[ -]*?){13,19}\b"
    ),

    "IP_ADDRESS": re.compile(
        r"\b(?:\d{1,3}\.){3}\d{1,3}\b"
    ),

    "SSN": re.compile(
        r"\b\d{3}-\d{2}-\d{4}\b"
    ),

    "PASSPORT": re.compile(
        r"\b[A-Z][0-9]{7}\b"
    ),
}


def redact_pii(text: str) -> Tuple[str, Dict[str, int]]:
    """
    Replaces detected PII with placeholders.

    Returns:
        redacted_text
        statistics
    """

    stats = {}

    for label, pattern in PII_PATTERNS.items():
        matches = pattern.findall(text)

        if matches:
            stats[label] = len(matches)

        text = pattern.sub(f"[REDACTED_{label}]", text)

    return text, stats
```

```python
message = """
Hi,

My name is John.

Email: john.doe@gmail.com
Phone: +1 987-654-3210
PAN: ABCDE1234F
Card: 4111 1111 1111 1111
IP: 192.168.1.15
"""

cleaned, stats = redact_pii(message)

print(cleaned)
print(stats)
```
