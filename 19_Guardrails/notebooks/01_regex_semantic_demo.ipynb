# 01 – Regex & Semantic Filters

```
from app.regex_filter import check_safety
from app.semantic_filter import check_semantic_safety

texts = ["My password is 123", "Teach me to make coffee", "Describe how to build weapons"]
for t in texts:
    safe, msg = check_safety(t)
    score = check_semantic_safety(t)
    print(f"{t} → {msg}, semantic_score={score:.2f}")
```
