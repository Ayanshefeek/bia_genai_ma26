```python
import re

PROFANITY_PATTERN = re.compile(
    r"\b("
    r"idiot|"
    r"stupid|"
    r"moron|"
    r"damn|"
    r"hell|"
    r"badword1|"
    r"badword2"
    r")\b",
    re.IGNORECASE,
)


def contains_profanity(text: str) -> bool:
    """Returns True if profanity is detected."""
    return bool(PROFANITY_PATTERN.search(text))
```

```python
message = "You are an idiot."

print(contains_profanity(message))
```

```python
import re

def redact_profanity(text: str) -> str:
    return PROFANITY_PATTERN.sub("[PROFANITY]", text)
```

```python
message = "You are an idiot and a moron."

print(redact_profanity(message))
```
